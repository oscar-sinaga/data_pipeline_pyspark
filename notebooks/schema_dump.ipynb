```python
import subprocess

pg_dump_path = r"C:\Program Files\PostgreSQL\17_binaries\bin\pg_dump.exe"

def get_schema_ddl_via_pg_dump(dbname, host, user, password, port, output_file):
    conn_str = f"postgresql://{user}:{password}@{host}:{port}/{dbname}"
    cmd = [
        pg_dump_path,
        f"--dbname={conn_str}",
        "--schema-only",
        "--no-owner",
        "--no-privileges"
    ]
    with open(output_file, "w") as f:
        # subprocess.run(cmd, stdout=f)

        result = subprocess.run(cmd, stdout=f, stderr=subprocess.PIPE, text=True)
        print("STDERR:", result.stderr)

# Contoh pemakaian
get_schema_ddl_via_pg_dump(
    dbname="pacbook",
    host="localhost",
    user="postgres",
    password="mypassword",
    port=5439,  # <--- ganti sesuai port kamu
    output_file="schema_dump.sql"
)

```
