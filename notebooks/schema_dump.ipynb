```python
import subprocess
import sys
import os

# Cari path ke folder src/
notebook_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(notebook_dir, '..'))
src_path = os.path.join(project_root, 'src')

# Tambahkan src ke sys.path
if src_path not in sys.path:
    sys.path.insert(0, src_path)

# Sekarang bisa import
from utils.helper import DB_USER, DB_PORT, DB_PASS, DB_NAME_STARTUP_INVESTMENTS

pg_dump_path = r"C:\Program Files\PostgreSQL\17_binaries\bin\pg_dump.exe"

def get_schema_ddl_via_pg_dump(dbname, host, user, password, port, output_file):
    conn_str = f"postgresql://{user}:{password}@{host}:{port}/{dbname}"
    cmd = [
        pg_dump_path,
        f"--dbname={conn_str}",
        "--schema-only",
        "--no-owner",
        "--no-privileges"
    ]
    with open(output_file, "w") as f:
        # subprocess.run(cmd, stdout=f)

        result = subprocess.run(cmd, stdout=f, stderr=subprocess.PIPE, text=True)
        print("STDERR:", result.stderr)

# Contoh pemakaian
get_schema_ddl_via_pg_dump(
    dbname=DB_NAME_STARTUP_INVESTMENTS,
    host="localhost",
    user=DB_USER,
    password=DB_PASS,
    port=DB_PORT,  # <--- ganti sesuai port kamu
    output_file="../schema/staging.sql"
)

```
